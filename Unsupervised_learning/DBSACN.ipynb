DBSCAN ---
Make cluster on basis if density (for non linear )
it is also used as a outlier detection


it takses 
1--min pts--------how many min pts to continue forming circles
2--eps(radius)
3--core pts--------no of pts>=min pts 
4--boundary pts---where circle ends
5--noise pts ------also known as outliers


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# to import make mooon dataset 
from sklearn.datasets import make_moons
x,y=make_moons(n_samples=250,noise=0.05)
```

```python
df={"data1":x[:,0],"data2":x[:,1]}
ds=pd.DataFrame(df)
```

```python
sns.scatterplot(x="data1",y="data2",data=ds)
```

```python
''' 
DBSCAN ----------------------
'''
from sklearn.cluster import DBSCAN
# keep in mind to remove the output (unsupervisedlearning --)
```

```python
db=DBSCAN(eps=0.2,min_samples=5)
pred=db.fit_predict(ds)
```

```python
ds["predict"]=pred
sns.scatterplot(x="data1",y="data2",data=ds,hue="predict")
```

# silhouette score 



--to check the number of cluster formed are right or not
--return -1 to 1

```python
#silhoutte score 
from sklearn.metrics import silhouette_score
silhouette_score(ds,pred)
```
