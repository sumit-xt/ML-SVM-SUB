```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
ds=pd.read_csv("flower_u.csv")
ds.head(3)
```

```python
sns.pairplot(data=ds)
```

```python
#to make dendogram---to find no of clusters
# then put that in angllo
import scipy.cluster.hierarchy as sc 
ln=sc.linkage(ds,method='single',metric='euclidean')

```

```python
#dendogram ---------
sc.dendrogram(ln)
plt.show()
```

here the least cutting like is for 2 

```python
''' 
now we will use anglomerative clustering 
'''
```

```python
from sklearn.cluster import AgglomerativeClustering
ac=AgglomerativeClustering(n_clusters=2,linkage="single")
```

```python
pred=ac.fit_predict(ds)
ds["predicted"]=pred
```

```python
sns.pairplot(data=ds,hue="predicted")
```
